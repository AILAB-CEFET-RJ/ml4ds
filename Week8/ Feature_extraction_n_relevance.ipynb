# Beer consumption as a classification issue
Let's try to predict whether "very little", "little", "much" or "very much" beer is drunk; A classification.

The limit of the classes is roughly based on the quartile distribution.

- I: under 22,000 Litres
- II: between 22,000 and 25,000 Litres
- III: between 25,000 and 29,000 Litres
- IV: More than 29,000 Litres

If all goes well, these classes are of comparable size (i.e. somewhat balanced).

### 1. Feature selection
Is the Date a useful feature for our goal? Would another be useful? How can you check?

We'll reuse the engineered calendar features (Month/Weekday sin-cos) and optionally add holidays. Think in terms of: (a) intuition-driven additions, (b) quick statistical filters, (c) model-driven selection.


```python
# Let's add the season, because why not
df['Season'] = df.Date.dt.month.map({1:1, 2:1, 3:2, 4:2, 5:2, 6:3, 7:3, 8:3, 9:4, 10:4, 11:4, 12:1})
```

```python
df['Month'] = df.Date.dt.month
```

```python
df["Month_sin"] = np.sin(2 * np.pi * df["Month"] / 12)
df["Month_cos"] = np.cos(2 * np.pi * df["Month"] / 12)

df["Weekday"] = df["Date"].dt.dayofweek + 1
df["Weekday_sin"] = np.cos(2 * np.pi * df["Weekday"] / 7)
df["Weekday_cos"] = np.cos(2 * np.pi * df["Weekday"] / 7)
```

### Filter-based relevance (mutual information)
Use mutual information to see non-linear associations. Higher values suggest a feature carries more signal about the class.


```python
from sklearn.feature_selection import mutual_info_classif

labels = df['Class']
feature_pool = ['AvgTemp', 'MinTemp', 'MaxTemp', 'Rainfall_mm', 'Weekend', 'Season', 'Month_sin', 'Month_cos', 'Weekday_sin', 'Weekday_cos']
X_pool = df[feature_pool]
mi = mutual_info_classif(X_pool, labels, random_state=42)
mi_df = pd.DataFrame({'feature': feature_pool, 'mutual_info': mi}).sort_values('mutual_info', ascending=False)
mi_df
```

```python
# Define feature sets so you can toggle experiments quickly
feature_sets = {
    "max_and_weekend": ["MaxTemp", "Weekend"],
    "seasonality": ["MaxTemp", "Weekend", "Month_sin", "Month_cos"],
    "calendar_plus": [c for c in ["MaxTemp", "Weekend", "Month_sin", "Month_cos", "Weekday_sin", "Weekday_cos", "HolyDay"] if c in df.columns]
}

selected_features = feature_sets["seasonality"]  # switch this to try other sets

labels = df.loc[:, 'Class']
data = df.loc[:, selected_features]

labels.value_counts().sort_index().plot(kind="bar", rot=0, title="Class counts")
plt.show()

print("Active feature set:", selected_features)
```

### Wrapper selection (RFE)
Recursive Feature Elimination wraps a model and drops the least useful features until a target count remains. Here we keep 4 features with Logistic Regression.


```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression

wrapper_features = ['MaxTemp', 'Weekend', 'Month_sin', 'Month_cos', 'Weekday_sin', 'Weekday_cos', 'Rainfall_mm']
X_wrapper = df[wrapper_features]
y_wrapper = labels

log_reg = LogisticRegression(max_iter=1000, random_state=42, multi_class='auto')
rfe = RFE(log_reg, n_features_to_select=4)
rfe.fit(X_wrapper, y_wrapper)

selected_by_rfe = [f for f, keep in zip(wrapper_features, rfe.support_) if keep]
print("RFE kept:", selected_by_rfe)
print("Ranking (1 = kept):", dict(zip(wrapper_features, rfe.ranking_)))
```

### Embedded + permutation importance
Tree-based models provide built-in importances; permutation checks how much shuffling a feature hurts performance on held-out data.


```python
from sklearn.inspection import permutation_importance
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

X_imp = df[wrapper_features]
y_imp = labels
X_imp_train, X_imp_test, y_imp_train, y_imp_test = train_test_split(
    X_imp, y_imp, test_size=0.25, random_state=0, stratify=y_imp
)

forest = RandomForestClassifier(n_estimators=200, random_state=42)
forest.fit(X_imp_train, y_imp_train)

fi = pd.Series(forest.feature_importances_, index=wrapper_features).sort_values(ascending=False)
print("RandomForest importances:\n", fi)

perm = permutation_importance(forest, X_imp_test, y_imp_test, n_repeats=10, random_state=42)
perm_df = pd.DataFrame({
    'feature': wrapper_features,
    'perm_importance': perm['importances_mean']
}).sort_values('perm_importance', ascending=False)
print("\nPermutation importance (held-out):")
print(perm_df)
```

# Function that plots the confusion matrix

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
import numpy as np

def plot_confusion_matrix(y_true, y_pred, model_name="Model"):
    """
    Plot a confusion matrix using seaborn heatmap.

    Parameters
    ----------
    y_true : array-like
        True class labels.
    y_pred : array-like
        Predicted class labels.
    model_name : str, optional
        Title for the plot (default is "Model").
    """
    # Compute confusion matrix
    confmat = confusion_matrix(y_true, y_pred)
    ticks = np.unique(np.concatenate((y_true, y_pred)))
    ticks.sort()

    # Create a seaborn heatmap
    plt.figure(figsize=(5, 5))
    sns.heatmap(
        confmat,
        annot=True,
        fmt="d",
        cmap="Blues",
        cbar=False,
        xticklabels=ticks,
        yticklabels=ticks,
        square=True
    )

    plt.title(f"Confusion Matrix - {model_name}")
    plt.xlabel("Predicted label")
    plt.ylabel("True label")
    plt.tight_layout()
    plt.show()
```

## 2. Baseline models for classification
Code a baseline model

```python
# splits the dataset in a training set and a test set
X_train, X_test, y_train, y_test = train_test_split(
    data, labels, test_size=0.25, random_state=0, stratify=labels
)

# give the size of the training and the test set again
print('shape of training set: ', X_train.shape)
print('shape of test set:     ', X_test.shape)
```

```python
def evaluate_feature_set(feature_names, model, model_name):
    X = df[feature_names]
    y = labels
    X_train_fs, X_test_fs, y_train_fs, y_test_fs = train_test_split(
        X, y, test_size=0.25, random_state=0, stratify=y
    )
    model.fit(X_train_fs, y_train_fs)
    y_pred_fs = model.predict(X_test_fs)
    acc = accuracy_score(y_test_fs, y_pred_fs)
    f1 = f1_score(y_test_fs, y_pred_fs, average='macro')
    print(f"{model_name} | Features: {feature_names}")
    print(f"Accuracy: {acc:.3f} | F1-macro: {f1:.3f}\n")
    plot_confusion_matrix(y_test_fs, y_pred_fs, model_name=model_name)
    return acc, f1
```

```python
feature_results = {}
for name, feats in feature_sets.items():
    model = LogisticRegression(max_iter=1000, random_state=42)
    acc, f1 = evaluate_feature_set(feats, model, f"LogReg ({name})")
    feature_results[name] = {'accuracy': acc, 'f1_macro': f1}

pd.DataFrame(feature_results).T
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 2.1 Choose a baseline classification model, fill in the blank
lr_model = LogisticRegression(max_iter=1000, random_state=42)

lr_model.fit(X_train, y_train)

y_pred = lr_model.predict(X_test)

lr_acc = accuracy_score(y_test, y_pred)
print('Accuracy:', lr_acc)

plot_confusion_matrix(y_test, y_pred, model_name="LogisticRegression")
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

knn_model = KNeighborsClassifier(n_neighbors=5)

knn_model.fit(X_train, y_train)

y_pred = knn_model.predict(X_test)

knn_acc = accuracy_score(y_test, y_pred)
print('Accuracy:', knn_acc)

plot_confusion_matrix(y_test, y_pred, model_name="KNeighborsClassifier")
```
